# 1. Data Structures

```python

def count_frequency(items):
    """
    Given a list of hashable items, return a dictionary where keys are the items
    and values are the number of times each item appears.
    Example: count_frequency(['a', 'b', 'a', 'c', 'b', 'a']) -> {'a': 3, 'b': 2, 'c': 1}
    """
    freq = {}
    for item in items:
        freq[item] = freq.get(item, 0) + 1
    return freq
    raise NotImplementedError("TODO: Implement count_frequency")
list=['a', 'b', 'a', 'c', 'b', 'a']
print(count_frequency(list))



def merge_dicts(d1, d2):
    """
    Merge two dictionaries. If a key exists in both, sum their values.
    Return a new dictionary.
    Example: merge_dicts({'a': 1, 'b': 2}, {'b': 3, 'c': 4}) -> {'a': 1, 'b': 5, 'c': 4}
    """
    merged = d1.copy()
    for key, value in d2.items():
        merged[key] = merged.get(key, 0) + value
    return merged

    raise NotImplementedError("TODO: Implement merge_dicts")
print(merge_dicts({'a': 1, 'b': 2}, {'b': 3, 'c': 4}))


def remove_duplicates(lst):
    """
    Return a new list with duplicates removed, preserving the original order.
    Example: remove_duplicates([1, 2, 1, 3, 2, 4]) -> [1, 2, 3, 4]
    """
    seen = set()
    result = []
    for item in lst:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result
    raise NotImplementedError("TODO: Implement remove_duplicates")
print(remove_duplicates([1, 2, 1, 3, 2, 4]))


def find_second_largest(nums):
    """
    Return the second largest number in a list of numbers.
    If the list has fewer than 2 unique numbers, return None.
    Example: find_second_largest([10, 5, 8, 10]) -> 8
    """
    largest = None
    second_largest = None

    for num in nums:
        if largest is None or num > largest:
            second_largest = largest
            largest = num
        elif num != largest and (second_largest is None or num > second_largest):
            second_largest = num

    return second_largest
    raise NotImplementedError("TODO: Implement find_second_largest")
print(find_second_largest([10, 5, 8, 10])  )
```



# 3. File Handling

```python
def read_file_lines(filepath):
    """
    Read a text file and return a list of lines without the trailing newline characters.
    If the file does not exist, return an empty list.
    """
    try:
        with open(filepath, 'r') as file:
            return file.read().splitlines()
    except FileNotFoundError:
        return []


def write_lines(filepath, lines):
    """
    Write a list of strings to a file, each on a new line.
    Overwrite the file if it already exists.
    """
    with open(filepath, 'w') as file:
        for line in lines:
            file.write(f"{line}\n")


def append_to_file(filepath, text):
    """
    Append a string to a file. Add a newline after the text.
    Create the file if it does not exist.
    """
    with open(filepath, 'a') as file:
        file.write(f"{text}\n")


def count_words_in_file(filepath):
    """
    Return the total number of words in a text file.
    Words are separated by whitespace.
    If the file does not exist, return 0.
    """
    try:
        with open(filepath, 'r') as file:
            return len(file.read().split())
    except FileNotFoundError:
        return 0


# Run the functions
if __name__ == "__main__":
    test_file = "test.txt"

    # Write initial content
    write_lines(test_file, ["Hello world", "Python file operations"])

    # Append content
    append_to_file(test_file, "Third line added")

    # Read back lines
    content = read_file_lines(test_file)
    print("File Content:", content)

    # Count words
    total_words = count_words_in_file(test_file)
    print("Total Words:", total_words)
```

# 4. Error Handling

```python
def safe_divide(a, b):
    """
    Return a / b. If b is zero, return None.
    If a or b is not a number, return None.
    """
    if type(a) not in (int, float) or type(b) not in (int, float):
        return None
    try:
        return a / b
    except ZeroDivisionError:
        return None


def read_file_safe(filepath):
    """
    Return the content of a file as a string.
    If the file does not exist, return None.
    """
    try:
        with open(filepath, 'r') as file:
            return file.read()
    except FileNotFoundError:
        return None


def parse_int(s):
    """
    Convert a string to an integer. If conversion fails, return None.
    Example: parse_int("123") -> 123, parse_int("abc") -> None
    """
    try:
        return int(s)
    except (ValueError, TypeError):
        return None


# Run and verify outputs
if __name__ == "__main__":
    print("--- Testing safe_divide ---")
    print("10 / 2:", safe_divide(10, 2))          # Expected: 5.0
    print("10 / 0:", safe_divide(10, 0))          # Expected: None
    print("'10' / 2:", safe_divide("10", 2))      # Expected: None

    print("\n--- Testing parse_int ---")
    print("parse_int('123'):", parse_int("123"))  # Expected: 123
    print("parse_int('abc'):", parse_int("abc"))  # Expected: None
    print("parse_int(None):", parse_int(None))   # Expected: None

    print("\n--- Testing read_file_safe ---")
    # Test reading non-existent file
    print("Non-existent file:", read_file_safe("missing.txt"))  # Expected: None

    # Test reading existent file
    with open("temp_test.txt", "w") as f:
        f.write("Hello, World!")

    print("Existent file:", read_file_safe("temp_test.txt"))     # Expected: Hello, World!
```

# 5. OOP Basics

```python
class BankAccount:
    """
    A simple bank account class.

    Attributes:
        owner (str): The name of the account owner.
        balance (float): The current balance (default 0).

    Methods:
        deposit(amount): Add amount to balance.
        withdraw(amount): Subtract amount from balance if sufficient funds.
                          Raise ValueError if insufficient funds.
        get_balance(): Return the current balance.
    """
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = float(balance)

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient funds.")
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        self.balance -= amount
        return self.balance

    def get_balance(self):
        return self.balance


class SavingsAccount(BankAccount):
    """
    A savings account that inherits from BankAccount and adds interest.

    Methods:
        add_interest(rate): Increase balance by rate percent.
                            Example: rate=5 means 5% interest.
    """
    def add_interest(self, rate):
        self.balance += self.balance * (rate / 100)
        return self.balance


class Rectangle:
    """
    A rectangle class.

    Attributes:
        width (float)
        height (float)

    Methods:
        area(): Return width * height.
        perimeter(): Return 2 * (width + height).
        is_square(): Return True if width == height, else False.
    """
    def __init__(self, width, height):
        self.width = float(width)
        self.height = float(height)

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def is_square(self):
        return self.width == self.height


# Run and verify outputs
if __name__ == "__main__":
    print("--- Testing BankAccount & SavingsAccount ---")
    acc = SavingsAccount("Alice", 100)
    print("Initial Balance:", acc.get_balance())

    acc.deposit(50)
    print("After $50 Deposit:", acc.get_balance())

    acc.withdraw(30)
    print("After $30 Withdrawal:", acc.get_balance())

    acc.add_interest(5)
    print("After 5% Interest:", acc.get_balance())

    print("\n--- Testing Rectangle ---")
    rect1 = Rectangle(10, 5)
    print(f"Rect (10x5) Area: {rect1.area()}, Perimeter: {rect1.perimeter()}, Is Square: {rect1.is_square()}")

    rect2 = Rectangle(4, 4)
    print(f"Rect (4x4) Area: {rect2.area()}, Perimeter: {rect2.perimeter()}, Is Square: {rect2.is_square()}")
```
