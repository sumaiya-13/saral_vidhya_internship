###Question 1: Correcting a Malformed JSON String

The following string is intended to be valid JSON but contains several errors. Identify all errors, correct them, and parse the JSON into a Python dictionary. Print the resulting dictionary.

````python
{
  "name": "Alice",
  "age": 30,
  "city": "New York",
  "hobbies": ["reading", "hiking",]
}
````

````python
import json

malformed_json = """{
  'name': 'Alice',
  age: 30,
  "city": "New York",
  "hobbies": ["reading", "hiking",]
}"""

corrected_json = """{
  "name": "Alice",
  "age": 30,
  "city": "New York",
  "hobbies": ["reading", "hiking"]
}"""

# Parse into a Python dictionary
data = json.loads(corrected_json)

# Print the resulting dictionary
print(data)



````





Single Quotes around Key and Value ('name': 'Alice'):

Issue: JSON standards strictly require double quotes (") for strings and object keys. Single quotes (') cause a syntax error.

Fix: Replaced 'name': 'Alice' with "name": "Alice".

Unquoted Object Key (age:):

Issue: All keys in JSON objects must be enclosed in double quotes. Unquoted keys are invalid JSON.

Fix: Replaced age: with "age":.

Trailing Comma in Array (["reading", "hiking",]):

Issue: Standard JSON does not permit trailing commas after the final element in an array or object.

Fix: Removed the extra comma after "hiking", making it ["reading", "hiking"].

#Question 2: Formatting a Python Dictionary to JSON

Convert the given Python dictionary to a JSON string with the following requirements:
- Indentation of 4 spaces
- Keys sorted alphabetically
- Non-ASCII characters preserved (ensure_ascii=False)
- Print the resulting JSON string.

````python
import json

data = {"name": "José", "age": 30, "city": "München", "hobbies": ["reading", "hiking"]}

# Convert dictionary to JSON string with required formatting
json_string = json.dumps(data, indent=4, sort_keys=True, ensure_ascii=False)

# Print the resulting JSON string
print(json_string)
````

## Question 3: Extracting and Formatting Nested JSON
Given the following JSON string representing a list of users:
1.   Extract the email of the user with id=2 and print it.
2.   Pretty-print the entire JSON with indentation of 2 spaces.



````python
import json

json_str = '''
[
  {"id": 1, "name": "Alice", "email": "alice@example.com"},
  {"id": 2, "name": "Bob", "email": "bob@example.com"},
  {"id": 3, "name": "Charlie", "email": "charlie@example.com"}
]
'''

# Parse the JSON string into a Python list
users = json.loads(json_str)

# 1. Extract and print the email of the user with id=2
email_id_2 = next((user["email"] for user in users if user["id"] == 2), None)
print(f"Email of user with id=2: {email_id_2}\n")

# 2. Pretty-print the entire JSON with 2 spaces indentation
pretty_json = json.dumps(users, indent=2)
print("Pretty-printed JSON:")
print(pretty_json)
````

## Question 4: Handling Duplicate Keys

The following JSON string contains duplicate keys:

```
{"a": 1, "b": 2, "a": 3}
```


1.   Explain what happens when you parse it with json.loads.
2.   Write code to parse it and merge duplicate keys by summing their values. Print the result.




````python
import json

dup_json = '{"a": 1, "b": 2, "a": 3}'

def parse_and_merge_duplicates(pairs):
    """
    Custom object_pairs_hook function.
    'pairs' is passed as a list of (key, value) tuples in the exact order found in JSON.
    Example: [('a', 1), ('b', 2), ('a', 3)]
    """
    merged = {}
    for key, value in pairs:
        merged[key] = merged.get(key, 0) + value
    return merged

# Use object_pairs_hook to catch raw key-value pairs before default overwriting happens
result = json.loads(dup_json, object_pairs_hook=parse_and_merge_duplicates)

print("Merged Dictionary:", result)
````

How we merged them:
To prevent json.loads() from dropping duplicate values during dictionary construction, we passed a custom hook function to the object_pairs_hook parameter.

json.loads() calls object_pairs_hook with an ordered list of key-value tuples directly as they are parsed: [('a', 1), ('b', 2), ('a', 3)].

Our function iterates through these tuples and builds a new dictionary using merged.get(key, 0) + value, accumulating and summing values for duplicate keys rather than overwriting them.

Question 5: Fixing JSON with Comments and Trailing Commas

The following string is not valid JSON because it contains // comments and a trailing comma. Remove the comments and the trailing comma, then parse it into a Python dictionary and print it.


```
{
  // user name
  "name": "Alice",
  "age": 30, // age in years
  "city": "New York",
}
```



````python
import json
import re

invalid_json = '''
{
  // user name
  "name": "Alice",
  "age": 30, // age in years
  "city": "New York",
}
'''

# Step 1: Remove single-line comments starting with //
json_no_comments = re.sub(r'//.*', '', invalid_json)

# Step 2: Remove trailing commas before closing braces } or brackets ]
clean_json = re.sub(r',\s*([\}\]])', r'\1', json_no_comments)

# Step 3: Parse the cleaned JSON string into a dictionary
data = json.loads(clean_json)

# Print the resulting dictionary
print(data)
````

Explanation: Describe the steps you took to clean the string before parsing.

---



Removed Single-Line Comments (// ...):

Regex: re.sub(r'//.*', '', invalid_json)

Action: Matches // and all subsequent text on that line up to the newline character, replacing it with an empty string. This strips out // user name and // age in years.

Removed Trailing Commas:

Regex: re.sub(r',\s*([\}\]])', r'\1', json_no_comments)

Action: Matches any comma , followed by optional whitespace (\s*) and a closing curly brace } or square bracket ]. It replaces the comma and whitespace with just the captured closing bracket (\1). This transforms "city": "New York", into "city": "New York".

Parsed into Dictionary:

Once comments and trailing commas are removed, the string conforms to valid JSON (RFC 8259) specifications, allowing json.loads() to parse it cleanly into a standard Python dictionary.

Question 6: Validating and Formatting a JSON String

Write a function validate_and_format(json_string) that:
- Attempts to parse the given JSON string.
- If parsing fails, returns "Invalid JSON".
- If parsing succeeds, returns the JSON string pretty-printed with indentation of 4 spaces and sorted keys.
- Test it with both a valid and an invalid JSON string.

````python
import json

def validate_and_format(json_string):
    """
    Attempts to parse a JSON string.
    If valid, returns the pretty-printed JSON (4-space indent, sorted keys).
    If invalid, returns "Invalid JSON".
    """
    try:
        parsed_data = json.loads(json_string)
        return json.dumps(parsed_data, indent=4, sort_keys=True)
    except (json.JSONDecodeError, TypeError):
        return "Invalid JSON"

# Test cases
valid = '{"b": 2, "a": 1}'
invalid = '{"a": 1, "b": 2,}'

print("--- Valid JSON Output ---")
print(validate_and_format(valid))

print("\n--- Invalid JSON Output ---")
print(validate_and_format(invalid))
````
